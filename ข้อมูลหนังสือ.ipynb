```python
from typing import Self
# 1.class book ข้อมูลหนังสือ
class Book:

  def __init__(self,
              book_isbn,
              book_title,
              book_author,
              price,
              category_id,
              shelf_locati,
              stock_qty,
              reserverd_qty,
              ):

              self.book_isbn = book_isbn
              self.book_title = book_title
              self.book_author = book_author
              self.price = price
              self.category_id = category_id
              self.shelf_locati = shelf_locati
              self.stock_qty = stock_qty
              self.reserverd_qty = reserverd_qty


```

```python
# 2.เอาข้อมูลจริงใส่
book1 = Book(
  book_isbn = "B101",
  book_title = "ผ่าพิภพไททัน / Attack on Titan",
  book_author = "Hajime Isayama",
  price = "90 บาท/เล่ม",
  category_id = "01",
  shelf_locati = "A-01",
  stock_qty = "34",
  reserverd_qty = "0", )


print("รหัส ISBN:", book1.book_isbn)
print("ชื่อเรื่อง:", book1.book_title)
print("ผู้แต่ง:", book1.book_author)
print("ราคา:", book1.price)
print("รหัสหมวดหมู่:",book1.category_id)

```

*   ค้นหาหนังสือ/หมวดหมู่
*   ตรวจสอบสต็อก/ตำแหน่ง


```python
import csv
import pandas as pd

# 1. Class Book
class Book:

    def __init__(
        self,
        book_isbn,
        book_title,
        book_author,
        price,
        category_id,
        sud_category_id,
        shelf_locati,
        stock_qty,
    ):
        self.book_isbn = book_isbn
        self.book_title = book_title
        self.book_author = book_author
        self.price = float(price)
        self.category_id = category_id
        self.sud_category_id = sud_category_id
        self.shelf_locati = shelf_locati
        self.stock_qty = int(stock_qty)
```

```python
# 2. สร้างไฟล์ CSV ของข้อมูลหนังสือการ์ตูนทั้งหมด 606 เล่ม
items = [
    ("ผ่าพิภพไททัน (Attack on Titan)", 34, "01", "01", "Hajime Isayama", 90, "A-01"),
    ("ดาบพิฆาตอสูร (Demon Slayer)", 23, "01", "01", "Koyoharu Gotouge", 90, "A-02"),
    ("วันพีช (One Piece)", 109, "01", "01", "Eiichiro Oda", 95, "A-03"),
    ("มหาเวทย์ผนึกมาร (Jujutsu Kaisen)", 28, "01", "01", "Gege Akutami", 90, "A-04"),
    ("โคนัน (Detective Conan)", 107, "01", "02", "Gosho Aoyama", 90, "A-05"),
    ("Spy x Family", 16, "01", "03", "Tatsuya Endo", 95, "A-06"),
    ("ไม่อยากถ้ายาก", 40, "01", "03", "Koji Kumeta", 75, "B-01"),
    ("ครอบครัวตึ๋งหนืด", 50, "01", "03", "Lee Bong-gi / Studio Mango", 165, "B-02"),
    ("เรื่องผีรอบโลก", 30, "02", "04", "Yeondoo Comics", 165, "B-03"),
    ("ล่าขุมทรัพย์สุดขอบฟ้า", 80, "02", "05", "Gomdori co.", 165, "B-04"),
    ("คุกกี้รัน เอาชีวิตรอด", 40, "02", "05", "Kim Kang-hwa", 165, "B-05"),
    ("คุกกี้รัน ผจญภัย", 49, "02", "05", "Song Do-soo", 165, "B-06"),
]

raw_data = []
series_num = 46

for title, total_vols, cat_id, sub_cat_id, author, price, shelf in items:
    for vol in range(1, total_vols + 1):
        isbn = f"B{series_num}{cat_id}{sub_cat_id}{vol:03d}"
        raw_data.append({
            "book_isbn": isbn,
            "book_title": f"{title} เล่ม {vol}",
            "book_author": author,
            "price": price,
            "category_id": cat_id,
            "sud_category_id": sub_cat_id,
            "shelf_locati": f"{shelf}",
            "stock_qty": 1 ,
        })

    series_num += 1

df = pd.DataFrame(raw_data)
df.to_csv("books_data_exact_606.csv", index=False, encoding="utf-8-sig")
```

```python
# 3. อ่านไฟล์ CSV
book_list = []

with open("books_data_exact_606.csv", mode="r", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)

    for row in reader:
        book = Book(
            book_isbn=row["book_isbn"],
            book_title=row["book_title"],
            book_author=row["book_author"],
            price=row["price"],
            category_id=row["category_id"],
            sud_category_id=row["sud_category_id"],
            shelf_locati=row["shelf_locati"],
            stock_qty=row["stock_qty"],
        )
        book_list.append(book)
```

```python
# 4. อ่านไฟล์ CSV เข้าสู่ List ของ Class Book
book_list = []
with open("books_data_exact_606.csv", mode="r", encoding="utf-8-sig") as file:
    reader = csv.DictReader(file)
    for row in reader:
        book = Book(
            book_isbn=row["book_isbn"],
            book_title=row["book_title"],
            book_author=row["book_author"],
            price=row["price"],
            category_id=row["category_id"],
            sud_category_id=row["sud_category_id"],
            shelf_locati=row["shelf_locati"],
            stock_qty=row["stock_qty"],
        )
        book_list.append(book)

print(f"โหลดข้อมูลสำเร็จทั้งหมด: {len(book_list)} เล่ม!")
```

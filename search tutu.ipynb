```python
import csv
import io
from urllib.request import urlopen

class BookSearch:
    def __init__(self, url: str):
        self.catalog = []
        # ดึงไฟล์ CSV ตรงจาก URL บน GitHub
        response = urlopen(url)
        csv_data = response.read().decode('utf-8-sig')

        # อ่านข้อมูล
        reader = csv.DictReader(io.StringIO(csv_data))
        for row in reader:
            self.catalog.append(row)

url = 'https://raw.githubusercontent.com/mooktuta/book-rental-system/refs/heads/main/books_data_exact_3030.csv'
search_system = BookSearch(url)
```

```python
import csv


# 1. นิยาม Class
class BookSearch:

    def __init__(self, filepath: str):
        self.catalog = []
        self.load_data(filepath)

    def load_data(self, filepath: str):
        try:
            with open(filepath, mode="r", encoding="utf-8-sig") as f:
                reader = csv.DictReader(f)
                self.catalog = [row for row in reader]
            print(
                "ระบบพร้อมใช้งาน: โหลดข้อมูลหนังสือเรียบร้อย"
                f" {len(self.catalog)} รายการ\n"
            )
        except Exception as e:
            print(f"เกิดข้อผิดพลาดในการโหลดไฟล์ CSV: {e}\n")

    def search_by_title(self, keyword: str):
        print(f"ผลการค้นหาชื่อหนังสือ: '{keyword}'")
        results = [
            b
            for b in self.catalog
            if keyword.lower() in b.get("ชื่อหนังสือ (Book Title)", "").lower()
        ]
        self._show_results(results)

    def search_by_isbn(self, isbn_code: str):
        print(f"ผลการค้นหารหัส/ISBN: '{isbn_code}'")
        results = [
            b
            for b in self.catalog
            if isbn_code.lower() in b.get("รหัส ISBN (ชื่อเรื่อง)", "").lower()
        ]
        self._show_results(results)

    def _show_results(self, results: list):
        if not results:
            print("ไม่พบข้อมูลหนังสือที่คุณต้องการค้นหา\n")
            return

        print(f"พบรายการหนังสือทั้งหมด {len(results)} รายการ")
        print("┌" + "─" * 76 + "┐")

        for i, b in enumerate(results, 1):
            isbn = b.get("รหัส ISBN (ชื่อเรื่อง)", "-")
            title = b.get("ชื่อหนังสือ (Book Title)", "-")
            author = b.get("ผู้แต่ง (Author)", "-")
            price = b.get("ราคาปก (บาท)", "-")
            shelf = b.get("ชั้นวาง (Shelf)", "-")
            status = b.get("สถานะ (Status)", "-")

            print(f"  [{i:02d}] {title}")
            print(
                f"       ISBN: {isbn:<14} ผู้แต่ง: {author:<22} ชั้นวาง:"
                f" {shelf}"
            )
            print(
                f"       ราคา: {price:>5} บาท             สถานะ: {status}"
            )

            if i < len(results):
                print("  " + "┈" * 36)

        print("└" + "─" * 76 + "┘\n")


# 2. สร้าง Object ตัวแปร
search_system = BookSearch("books_data_exact_3030.csv")

```

```python
# 3. ทดสอบค้นหา ค้นหาด้วยชื่อ
search_system.search_by_title("ผ่าพิภพไททัน")
```

```python
# ตัวอย่างค้นหาด้วย "รหัส/ISBN"
search_system.search_by_isbn("B470101001")
```

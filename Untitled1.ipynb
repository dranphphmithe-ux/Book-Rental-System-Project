```python
import random
from typing import List, Optional

class Customer:
    def __init__(self, customer_id: int, name: str, phone: str, email: str):
        self.customer_id = customer_id
        self.name = name
        self.phone = phone
        self.email = email
        self.points = 0


class CustomerDatabase:
    def __init__(self):
        self.customers: List[Customer] = []

    def generate_mock_customers(self, count: int = 300) -> None:
        """จำลองข้อมูลลูกค้าจำนวน 300 คนเข้าสู่ระบบ"""
        first_names = ["กิตติ", "สมชาย", "วิภา", "ณัฐพงษ์", "อารียา", "ธนกฤต", "ศิริพร", "ชยพล", "กนกวรรณ", "ปรีชา"]
        last_names = ["ใจดี", "รักเรียน", "มีสุข", "เจริญรุ่ง", "มงคล", "สุวรรณ", "วงศ์ใหญ่", "สายอ่าน", "เกียรติไพศาล", "มั่นคง"]

        self.customers.clear()
        for i in range(1, count + 1):
            fname = random.choice(first_names)
            lname = random.choice(last_names)
            name = f"{fname} {lname} ({i})"
            phone = f"08{random.randint(10000000, 99999999)}"
            email = f"user{i}@example.com"

            customer = Customer(customer_id=1000 + i, name=name, phone=phone, email=email)
            self.customers.append(customer)

        print(f"สร้างข้อมูลลูกค้าจำลองสำเร็จ: {len(self.customers)} คน")

    def show_all_customers(self, limit: int = 5) -> None:
        """แสดงรายชื่อลูกค้า"""
        print(f"\n================ รายชื่อลูกค้าในระบบ (ทั้งหมด {len(self.customers)} คน) ================")
        for idx, c in enumerate(self.customers[:limit], 1):
            print(f"{idx}. ID: {c.customer_id} | ชื่อ: {c.name} | เบอร์โทร: {c.phone} | อีเมล: {c.email}")
        if len(self.customers) > limit:
            print(f"... และอีก {len(self.customers) - limit} คน ...")
        print("========================================================================\n")
```

```python
class Book:
    def __init__(self, isbn: str, title: str, author: str, price: float,
                 rental_rate_per_day: float, deposit_amount: float,
                 category_id: int, shelf_location: str, stock_quantity: int):
        self.isbn = isbn
        self.title = title
        self.author = author
        self.price = price
        self.rental_rate_per_day = rental_rate_per_day
        self.deposit_amount = deposit_amount
        self.category_id = category_id
        self.shelf_location = shelf_location
        self.stock_quantity = stock_quantity

    @property
    def is_available(self) -> bool:
        return self.stock_quantity > 0

    def rent_out(self, qty: int = 1) -> bool:
        if self.stock_quantity >= qty:
            self.stock_quantity -= qty
            return True
        return False


class BookInventory:
    def __init__(self):
        self.books: List[Book] = []

    def generate_mock_books(self, count: int = 600) -> None:
        """จำลองข้อมูลหนังสือ 600 เล่ม"""
        categories = ["นิยายแฟนตาซี", "มังงะ/การ์ตูน", "บริหาร/ธุรกิจ", "เทคโนโลยี"]
        self.books.clear()

        for i in range(1, count + 1):
            cat_id = (i % 4) + 1
            price = float(random.choice([150, 200, 250, 300, 350, 400]))
            rental_rate = round(price * 0.10, 2)
            deposit = round(price * 0.50, 2)

            book = Book(
                isbn=f"978-616-{i:05d}",
                title=f"หนังสือ{categories[cat_id-1]} เล่มที่ {i}",
                author=f"นักเขียนท่านที่ {(i % 50) + 1}",
                price=price,
                rental_rate_per_day=rental_rate,
                deposit_amount=deposit,
                category_id=cat_id,
                shelf_location=f"Shelf-{(i // 50) + 1:02d}",
                stock_quantity=random.randint(2, 5)
            )
            self.books.append(book)

        print(f"สร้างข้อมูลหนังสือจำลองสำเร็จ: {len(self.books)} เล่ม")

    def get_available_books(self) -> List[Book]:
        return [b for b in self.books if b.is_available]
```

```python
from datetime import datetime, timedelta

class RentalItem:
    def __init__(self, item_id: int, book: Book, rental_days: int):
        self.item_id = item_id
        self.book = book
        self.rental_days = rental_days
        self.rental_fee = round(book.rental_rate_per_day * rental_days, 2)
        self.deposit = book.deposit_amount


class RentalOrder:
    def __init__(self, order_id: int, customer: Customer):
        self.order_id = order_id
        self.customer = customer
        self.rental_date = datetime.now()
        self.due_date: Optional[datetime] = None
        self.items: List[RentalItem] = []
        self.total_rental_fee = 0.0
        self.total_deposit = 0.0
        self.total_amount = 0.0

    def add_rental_item(self, book: Book, rental_days: int) -> bool:
        if book.rent_out(1):
            item_id = len(self.items) + 1
            item = RentalItem(item_id, book, rental_days)
            self.items.append(item)

            item_due = self.rental_date + timedelta(days=rental_days)
            if self.due_date is None or item_due > self.due_date:
                self.due_date = item_due

            self._calculate_totals()
            return True
        return False

    def _calculate_totals(self) -> None:
        self.total_rental_fee = round(sum(item.rental_fee for item in self.items), 2)
        self.total_deposit = round(sum(item.deposit for item in self.items), 2)
        self.total_amount = round(self.total_rental_fee + self.total_deposit, 2)
```

```python
if __name__ == "__main__":
    # 1. โหลดข้อมูลลูกค้า 300 คน และ หนังสือ 600 เล่ม
    customer_db = CustomerDatabase()
    customer_db.generate_mock_customers(count=300)

    inventory = BookInventory()
    inventory.generate_mock_books(count=600)

    # แสดงตัวอย่างข้อมูล
    customer_db.show_all_customers(limit=3)

    # 2. จำลองการเช่าหนังสือจริงโดยสุ่มลูกค้ามาเช่าหนังสือ
    print(">>> เริ่มการจำลองทำรายการเช่าหนังสือ <<<")
    orders: List[RentalOrder] = []

    # สุ่มให้ลูกค้า 50 คนจาก 300 คนเข้ามาเช่าหนังสือ
    active_customers = random.sample(customer_db.customers, 50)

    for idx, cust in enumerate(active_customers, 1):
        order = RentalOrder(order_id=5000 + idx, customer=cust)

        # สุ่มเช่าหนังสือ 1-3 เล่มต่อคน
        available_books = inventory.get_available_books()
        if not available_books:
            break

        books_to_rent = random.sample(available_books, min(random.randint(1, 3), len(available_books)))
        for b in books_to_rent:
            order.add_rental_item(b, rental_days=random.choice([3, 5, 7]))

        orders.append(order)

    # 3. แสดงสรุปผลการจำลอง
    print(f"\nทำรายการเช่าสำเร็จทั้งหมด: {len(orders)} ออเดอร์")
    sample_order = orders[0]
    print(f"\n--- ตัวอย่างออเดอร์ที่ 1 ---")
    print(f"รหัสออเดอร์: #{sample_order.order_id}")
    print(f"ผู้เช่า: {sample_order.customer.name} (ID: {sample_order.customer.customer_id})")
    print(f"จำนวนหนังสือที่เช่า: {len(sample_order.items)} เล่ม")
    print(f"ยอดค่าเช่า: {sample_order.total_rental_fee} บาท | เงินมัดจำ: {sample_order.total_deposit} บาท")
    print(f"ยอดรวมสุทธิ: {sample_order.total_amount} บาท")
```

```python
import random
import csv
import sqlite3
from datetime import datetime, timedelta
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ตั้งค่าการแสดงผลกราฟภาษาไทยและ Style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
plt.rcParams['font.sans-serif'] = ['Tahoma', 'Angsana New', 'DejaVu Sans'] # รองรับฟอนต์
plt.rcParams['axes.unicode_minus'] = False


# ==========================================
# 1. STRUCTURAL PROGRAMMING & CLASSES
# ==========================================

# --- 1.1 ออกแบบ Class (3 Classes หลัก) ---
class Book:
    def __init__(self, isbn: str, title: str, category: str, price: float, stock_quantity: int):
        self.isbn = isbn
        self.title = title
        self.category = category
        self.price = price
        self.rental_rate_per_day = round(price * 0.10, 2)  # ค่าเช่า 10% ของราคาหนังสือ
        self.deposit_amount = round(price * 0.50, 2)      # ค่ามัดจำ 50%
        self.stock_quantity = stock_quantity

    def rent(self) -> bool:
        if self.stock_quantity > 0:
            self.stock_quantity -= 1
            return True
        return False

    def return_book(self) -> None:
        self.stock_quantity += 1


class Customer:
    def __init__(self, customer_id: int, name: str, phone: str, member_tier: str = "Member"):
        self.customer_id = customer_id
        self.name = name
        self.phone = phone
        self.member_tier = member_tier
        self.points = 0


class RentalTransaction:
    def __init__(self, transaction_id: int, customer: Customer, book: Book, rental_days: int, rental_date: datetime):
        self.transaction_id = transaction_id
        self.customer_id = customer.customer_id
        self.isbn = book.isbn
        self.rental_days = rental_days
        self.rental_date = rental_date
        self.due_date = rental_date + timedelta(days=rental_days)

        # คำนวณยอดเงิน
        self.rental_fee = round(book.rental_rate_per_day * rental_days, 2)
        self.deposit = book.deposit_amount
        self.discount = self.calculate_discount(customer.member_tier, self.rental_fee)
        self.net_rental_fee = round(self.rental_fee - self.discount, 2)
        self.total_paid = round(self.net_rental_fee + self.deposit, 2)

    @staticmethod
    def calculate_discount(tier: str, fee: float) -> float:
        discount_rates = {"Member": 0.0, "Silver": 0.05, "Gold": 0.10, "VIP": 0.15}
        return round(fee * discount_rates.get(tier, 0.0), 2)


# --- 1.2 Function ช่วยทำงาน (3 Functions) ---

# Function 1: คำนวณค่าปรับเกินกำหนด (มี parameter, return, default argument)
def calculate_overdue_fine(rental_days: int, actual_returned_days: int, fine_rate_per_day: float = 15.0) -> float:
    """คำนวณค่าปรับเมื่อคืนหนังสือเกินกำหนด (Default fine = 15 บาท/วัน)"""
    if actual_returned_days > rental_days:
        overdue_days = actual_returned_days - rental_days
        return round(overdue_days * fine_rate_per_day, 2)
    return 0.0

# Function 2: ตรวจสอบและอัปเดตสิทธิ์ส่วนลดสมาชิก
def get_member_tier_by_points(points: int) -> str:
    if points >= 500: return "VIP"
    elif points >= 300: return "Gold"
    elif points >= 100: return "Silver"
    return "Member"

# Function 3: จัดรูปแบบวันที่ให้อยู่ในสตรีมที่อ่านง่าย
def format_datetime_str(dt: datetime, date_format: str = "%Y-%m-%d %H:%M") -> str:
    return dt.strftime(date_format)


# --- 1.3 จำลองการรับข้อมูลด้วย Loop & Random Module ---

# เตรียม Master Data: หนังสือ 600 เล่ม และ ลูกค้า 300 คน
categories = ["มังงะ/การ์ตูน", "นิยายแฟนตาซี", "บริหาร/ธุรกิจ", "เทคโนโลยี/คอมพิวเตอร์", "พัฒนาตนเอง"]
book_pool = []
for i in range(1, 601):
    cat = random.choice(categories)
    price = float(random.choice([100, 150, 200, 250, 300, 450]))
    book_pool.append(Book(f"978-616-{i:05d}", f"หนังสือ {cat} เล่มที่ {i}", cat, price, stock_quantity=random.randint(3, 10)))

first_names = ["กิตติ", "สมชาย", "วิภา", "ณัฐพงษ์", "อารียา", "ธนกฤต", "ศิริพร", "ชยพล", "กนกวรรณ", "ปรีชา"]
last_names = ["ใจดี", "รักเรียน", "มีสุข", "เจริญรุ่ง", "มงคล", "สุวรรณ", "วงศ์ใหญ่", "สายอ่าน", "เกียรติไพศาล"]
customer_pool = []
for i in range(1, 301):
    name = f"{random.choice(first_names)} {random.choice(last_names)}"
    phone = f"08{random.randint(10000000, 99999999)}"
    tier = random.choice(["Member", "Silver", "Gold", "VIP"])
    customer_pool.append(Customer(1000 + i, name, phone, tier))

# LOOP จำลองธุรกรรมการเช่าต่อเนื่องทีละรายการ (อย่างน้อย 300 รายการ)
transactions_list = []
start_date = datetime(2026, 1, 1)

for i in range(1, 350): # จำลอง 350 รายการเช่า
    # สุ่มข้อมูลธุรกรรมแต่ละรอบ
    cust = random.choice(customer_pool)
    available_books = [b for b in book_pool if b.stock_quantity > 0]
    selected_book = random.choice(available_books)

    rental_days = random.choice([3, 5, 7, 10])
    tx_date = start_date + timedelta(days=random.randint(0, 180), hours=random.randint(9, 18))

    # ทำรายการเช่าและตัดสต็อก
    if selected_book.rent():
        tx = RentalTransaction(
            transaction_id=5000 + i,
            customer=cust,
            book=selected_book,
            rental_days=rental_days,
            rental_date=tx_date
        )

        # บันทึกเป็น Dictionary เพื่อนำไปใช้ต่อ
        transactions_list.append({
            "transaction_id": tx.transaction_id,
            "customer_id": tx.customer_id,
            "customer_name": cust.name,
            "member_tier": cust.member_tier,
            "isbn": tx.isbn,
            "book_title": selected_book.title,
            "category": selected_book.category,
            "rental_days": tx.rental_days,
            "rental_date": format_datetime_str(tx.rental_date),
            "due_date": format_datetime_str(tx.due_date),
            "rental_fee": tx.rental_fee,
            "discount": tx.discount,
            "net_rental_fee": tx.net_rental_fee,
            "deposit": tx.deposit,
            "total_paid": tx.total_paid
        })

print(f"จำลองธุรกรรมเช่าหนังสือเรียบร้อยแล้วทั้งหมด: {len(transactions_list)} รายการ")


# --- 1.4 บันทึกข้อมูลลง CSV ---
csv_filename = "book_rentals.csv"
fieldnames = list(transactions_list[0].keys())

with open(csv_filename, mode='w', encoding='utf-8-sig', newline='') as f:
    writer = csv.DictWriter(f, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(transactions_list)

print(f"บันทึกไฟล์ CSV สำเร็จ: {csv_filename}")


# --- 1.5 สร้างและ Insert ลงฐานข้อมูล SQLite3 (แยก 2 ตาราง + JOIN) ---
conn = sqlite3.connect("book_rental_db.db")
cursor = conn.cursor()

# สร้างตาราง Customers
cursor.execute("""
CREATE TABLE IF NOT EXISTS customers (
    customer_id INTEGER PRIMARY KEY,
    name TEXT,
    phone TEXT,
    member_tier TEXT
)
""")

# สร้างตาราง Transactions
cursor.execute("""
CREATE TABLE IF NOT EXISTS transactions (
    transaction_id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    isbn TEXT,
    book_title TEXT,
    category TEXT,
    rental_days INTEGER,
    rental_date TEXT,
    net_rental_fee REAL,
    deposit REAL,
    total_paid REAL,
    FOREIGN KEY (customer_id) REFERENCES customers (customer_id)
)
""")

# Insert ข้อมูล
cursor.executemany("""
INSERT OR REPLACE INTO customers VALUES (?, ?, ?, ?)
""", [(c.customer_id, c.name, c.phone, c.member_tier) for c in customer_pool])

cursor.executemany("""
INSERT OR REPLACE INTO transactions VALUES (?, ?, ?, ?, ?, ?, ?, ?, ?, ?)
""", [(t["transaction_id"], t["customer_id"], t["isbn"], t["book_title"], t["category"],
       t["rental_days"], t["rental_date"], t["net_rental_fee"], t["deposit"], t["total_paid"]) for t in transactions_list])

conn.commit()
print("สร้างและนำเข้าข้อมูลสู่ SQLite Database สำเร็จ")


# ==========================================
# 2. DATA ANALYSIS WITH PANDAS
# ==========================================
df = pd.read_csv("book_rentals.csv")

print("\n--- 2.1 สำรวจข้อมูลเบื้องต้นด้วย .info() และ .describe() ---")
df.info()
print("\nค่าสถิติเชิงพรรณนา:")
display(df.describe())

print("\n--- 2.2 สรุปข้อมูลด้วย .groupby() + .agg() ---")
category_summary = df.groupby('category').agg(
    total_rentals=('transaction_id', 'count'),
    total_revenue=('net_rental_fee', 'sum'),
    avg_rental_days=('rental_days', 'mean')
).reset_index()
display(category_summary)

print("\n--- 2.3 จัดลำดับ หา Top 5 ลูกค้าที่ใช้จ่ายสูงสุด (.sort_values()) ---")
top_5_customers = df.groupby(['customer_id', 'customer_name', 'member_tier'])['total_paid'].sum().reset_index()
top_5_customers = top_5_customers.sort_values(by='total_paid', ascending=False).head(5)
display(top_5_customers)


# ==========================================
# 3. SQL QUERIES WITH SQLITE3
# ==========================================
print("\n--- 3.1 SQL Queries (WHERE, ORDER BY) ---")
q1 = """
SELECT transaction_id, book_title, category, net_rental_fee
FROM transactions
WHERE net_rental_fee > 100
ORDER BY net_rental_fee DESC
LIMIT 5;
"""
df_q1 = pd.read_sql_query(q1, conn)
display(df_q1)

print("\n--- 3.2 SQL Query (GROUP BY + Aggregate) ---")
q2 = """
SELECT category, COUNT(transaction_id) as rental_count, AVG(net_rental_fee) as avg_fee
FROM transactions
GROUP BY category
ORDER BY rental_count DESC;
"""
df_q2 = pd.read_sql_query(q2, conn)
display(df_q2)

print("\n--- 3.3 SQL Query (JOIN ตาราง) ---")
q3 = """
SELECT t.transaction_id, c.name, c.member_tier, t.book_title, t.net_rental_fee
FROM transactions t
INNER JOIN customers c ON t.customer_id = c.customer_id
WHERE c.member_tier = 'VIP'
LIMIT 5;
"""
df_q3 = pd.read_sql_query(q3, conn)
display(df_q3)

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# --------------------------------------------------
# กราฟที่ 1: Bar Chart (แนวตั้ง)
# - แกน X: หมวดหมู่หนังสือ (category)
# - แกน Y: รายได้รวมสุทธิ (total_revenue)
# - วัตถุประสงค์: เปรียบเทียบรายได้รวมที่ร้านได้รับจากหนังสือแต่ละหมวดหมู่
# --------------------------------------------------
sns.barplot(
    data=category_summary,
    x='category',
    y='total_revenue',
    ax=axes[0],
    palette='Blues_d'
)
axes[0].set_title('1. รายได้รวมจำแนกตามหมวดหมู่หนังสือ (Bar Chart)', fontsize=12, fontweight='bold')
axes[0].set_xlabel('หมวดหมู่หนังสือ (Category)', fontsize=10)
axes[0].set_ylabel('รายได้รวม (บาท) (Total Revenue)', fontsize=10)
axes[0].tick_params(axis='x', rotation=30)  # เอียงข้อความแกน X เพื่อให้อ่านง่าย


# --------------------------------------------------
# กราฟที่ 2: Histogram + KDE
# - แกน X: จำนวนวันที่เช่าหนังสือ (rental_days)
# - แกน Y: ความถี่ / จำนวนครั้งที่มีการเช่า (Frequency / Count)
# - วัตถุประสงค์: ดูการกระจายตัวของระยะเวลาที่ลูกค้าชอบเช่าหนังสือ (กี่วันมากที่สุด)
# --------------------------------------------------
sns.histplot(
    data=df,
    x='rental_days',
    bins=10,
    kde=True,
    ax=axes[1],
    color='teal'
)
axes[1].set_title('2. การกระจายตัวของจำนวนวันที่เช่าหนังสือ (Histogram)', fontsize=12, fontweight='bold')
axes[1].set_xlabel('จำนวนวันที่เช่า (วัน) (Rental Days)', fontsize=10)
axes[1].set_ylabel('จำนวนธุรกรรมการเช่า (ครั้ง) (Transaction Count)', fontsize=10)


# --------------------------------------------------
# กราฟที่ 3: Scatter Plot (จุดกระจาย)
# - แกน X: ค่ามัดจำหนังสือ (deposit)
# - แกน Y: ค่าเช่าสุทธิ (net_rental_fee)
# - สีของจุด (Hue): ระดับสมาชิกของลูกค้า (member_tier)
# - วัตถุประสงค์: วิเคราะห์ความสัมพันธ์ระหว่างเงินมัดจำและค่าเช่า แยกตามระดับสมาชิก
# --------------------------------------------------
sns.scatterplot(
    data=df,
    x='deposit',
    y='net_rental_fee',
    hue='member_tier',
    ax=axes[2],
    s=80,
    palette='Set2'
)
axes[2].set_title('3. ความสัมพันธ์ระหว่างมัดจำและค่าเช่าสุทธิ (Scatter Plot)', fontsize=12, fontweight='bold')
axes[2].set_xlabel('เงินมัดจำหนังสือ (บาท) (Deposit)', fontsize=10)
axes[2].set_ylabel('ค่าเช่าสุทธิหลังหักส่วนลด (บาท) (Net Rental Fee)', fontsize=10)
axes[2].legend(title='ระดับสมาชิก (Tier)') # กำกับชื่อกล่องสัญลักษณ์


# จัดระยะห่างระหว่างกราฟให้ออกมาสวยงาม ไม่ซ้อนทับกัน
plt.tight_layout()
```

```python
import csv
import io
from urllib.request import urlopen


class BookSearch:

  def __init__(self, url: str):
    self.catalog = []
    response = urlopen(url)
    csv_data = response.read().decode('utf-8-sig')
    reader = csv.DictReader(io.StringIO(csv_data))
    for row in reader:
      self.catalog.append(row)

  def search_by_title(self, keyword: str):
    print(f"ผลการค้นหาชื่อหนังสือ: '{keyword}'")
    results = [
        b
        for b in self.catalog
        if keyword.lower() in b.get('book_title', '').lower()
    ]
    self._show_results(results)

  def search_by_isbn(self, isbn_code: str):
    print(f"ผลการค้นหารหัส/ISBN: '{isbn_code}'")
    results = [
        b
        for b in self.catalog
        if isbn_code.lower() in b.get('book_isbn', '').lower()
    ]
    self._show_results(results)

  def search_by_category(self, category_id: str):
    print(f"ผลการค้นหาหมวดหมู่ ID: '{category_id}'")
    results = [
        b
        for b in self.catalog
        if str(category_id) == str(b.get('sud_category_id', ''))
    ]
    self._show_results(results)

  def _show_results(self, results: list):
    if not results:
      print('ไม่พบข้อมูลหนังสือที่คุณต้องการค้นหา\n')
      return

    print(f'พบรายการหนังสือทั้งหมด {len(results)} รายการ')
    print('┌' + '─' * 76 + '┐')

    for i, b in enumerate(results, 1):
      isbn = b.get('book_isbn', '-')
      title = b.get('book_title', '-')
      author = b.get('book_author', '-')
      price = b.get('price', '-')
      shelf = b.get('shelf_locati', '-')
      stock = b.get('stock_qty', '0')

      print(f'  [{i:02d}] {title}')
      print(
          f'       ISBN: {isbn:<14} ผู้แต่ง: {author:<22} ชั้นวาง:'
          f' {shelf}'
      )
      print(
          f'       ราคา: {price:>3} บาท            จำนวนคงเหลือ: {stock} เล่ม'
      )

      if i < len(results):
        print('  ' + '┈' * 36)

    print('└' + '─' * 76 + '┘\n')
```

```python
# ตัวอย่างค้นหาด้วย "ชื่อหนังสือ"
search_system.search_by_title("วันพีช")
```

```python
# ตัวอย่างค้นหาด้วย "รหัส ISBN"
search_system.search_by_isbn("B460101109")
```

```python
# 3. ตัวอย่างค้นหาด้วย "ID หมวดหมู่"
search_system.search_by_category("1")
```
